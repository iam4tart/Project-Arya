```python
import os
import pandas as pd
import librosa
import librosa.display
import matplotlib.pyplot as plt
import numpy as np

def extract_spectrogram(file_path, output_path):
    # Load audio file
    y, sr = librosa.load(file_path, sr=None)

    # Compute spectrogram
    spectrogram = np.abs(librosa.stft(y))

    # Plot and save spectrogram
    plt.figure(figsize=(10, 4))
    librosa.display.specshow(librosa.amplitude_to_db(spectrogram, ref=np.max), sr=sr, x_axis='time', y_axis='log')
    plt.colorbar(format='%+2.0f dB')
    plt.title('Spectrogram')
    plt.savefig(output_path)
    plt.close()

def get_files_with_features(path):
    data_ = []
    count = {}
    files = os.listdir(path)
    batch_size = 10  # Adjust batch size based on memory constraints
    for i in range(0, len(files), batch_size):
        batch_files = files[i:i+batch_size]
        for file in batch_files:
            labels = file.split("-")[:3]
            data = {
                "keyword": labels[0],
                "stress": labels[2],
                "environment": labels[1],
                "path": os.path.join(path, file)
            }
            for label in labels:
                if label not in count.keys():
                    count[label] = 1
                else:
                    count[label] += 1
            
            # Extract spectrogram image
            spectrogram_path = os.path.splitext(data['path'])[0] + ".png"
            extract_spectrogram(data['path'], spectrogram_path)
            data['spectrogram_path'] = spectrogram_path

            data_.append(data)

    if data_:
        df = pd.DataFrame(data_)
    else:
        df = pd.DataFrame(columns=['keyword', 'stress', 'environment', 'path', 'spectrogram_path'])
    return data_, count, df

# Example usage
path = 'EmoSpeech/'
dataset_with_features, count, df_with_features = get_files_with_features(path)
print(df_with_features.head())
```
