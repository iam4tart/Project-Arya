```python
import os
import shutil

def merge_folders(source_folders, destination_folder):
    # Create the destination folder if it doesn't exist
    if not os.path.exists(destination_folder):
        os.makedirs(destination_folder)
    
    # Iterate over each source folder
    for source_folder in source_folders:
        # Iterate over the files and subfolders in the source folder
        for item in os.listdir(source_folder):
            source_path = os.path.join(source_folder, item)
            destination_path = os.path.join(destination_folder, item)
            
            # If the item is a file, copy it to the destination folder
            if os.path.isfile(source_path):
                shutil.copy2(source_path, destination_path)
            
            # If the item is a folder, merge its contents recursively
            elif os.path.isdir(source_path):
                merge_folders([source_path], destination_path)

# Example usage:
source_directory = '/Users/pabitraghosh/Downloads/augmented'
destination_folder = '/Users/pabitraghosh/Downloads/destination_folder'

# Specify the folders to merge
source_folders = [
    os.path.join(source_directory, 'EmoSpeech 2'),
    os.path.join(source_directory, 'pitchshift'),
    os.path.join(source_directory, 'noise'),
    os.path.join(source_directory, 'timestretch')
]

# Merge the folders into a single destination folder
merge_folders(source_folders, destination_folder)
```
